# 06 模型解释与可视化

展示扫描路径、热点图和基于 permutation importance 的解释结果。SHAP 可在安装 `gaze-toolkit[explain]` 后继续接入。

```python
import matplotlib.pyplot as plt

from gaze_toolkit.datasets import simulate_gaze_recording, simulate_intent_dataset
from gaze_toolkit.events import attach_events
from gaze_toolkit.modeling import permutation_feature_importance, train_model
from gaze_toolkit.visualization import plot_heatmap, plot_scanpath

recording = attach_events(simulate_gaze_recording(seed=8))
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
plot_scanpath(recording, ax=axes[0])
plot_heatmap(recording, ax=axes[1])
plt.tight_layout()
```

```python
dataset = simulate_intent_dataset(num_sessions=24)
result = train_model(dataset, target='intent_label')
importance = permutation_feature_importance(result, dataset[result.feature_names], dataset['intent_label'])
importance.head(10)
```
