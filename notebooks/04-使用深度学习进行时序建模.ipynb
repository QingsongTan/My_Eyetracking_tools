# 04 使用深度学习进行时序建模

MVP 包默认不捆绑深度模型，但已经把时序输入准备步骤独立出来。下面先把眼动记录转成固定长度序列，后续可以直接接 LSTM、TCN 或 Transformer。

```python
import numpy as np

from gaze_toolkit.datasets import simulate_gaze_recording

recordings = [simulate_gaze_recording(style='careful' if i % 2 == 0 else 'skim', seed=i) for i in range(8)]
window_length = 60

def to_sequence(recording):
    frame = recording.samples[['x', 'y', 'pupil']].ffill().fillna(0.0)
    return frame.iloc[:window_length].to_numpy()

X = np.stack([to_sequence(recording) for recording in recordings])
y = np.array([1 if recording.metadata['intent_label'] == 'careful' else 0 for recording in recordings])
X.shape, y
```
